```python
import networkx as nx
import matplotlib.pyplot as plt

# -------------------------------------------------------------------
# Helper function to build and save a graph
# -------------------------------------------------------------------
def draw_graph(nodes, edges, out_file):
    G = nx.DiGraph()
    G.add_nodes_from(nodes)
    G.add_edges_from(edges)

    plt.figure(figsize=(10, 8))
    pos = nx.spring_layout(G, k=0.8, seed=42)

    # Draw nodes with larger size and slightly transparent dark blue
    nx.draw_networkx_nodes(G, pos, node_color="lightblue", node_size=5000, alpha=0.95)

    # Draw labels in white with larger font
    nx.draw_networkx_labels(G, pos, font_size=12, font_color="black", font_weight='bold')

    # Draw edges with slightly darker color and thicker lines
    nx.draw_networkx_edges(G, pos, arrows=True, arrowstyle='-|>', 
                           arrowsize=25, edge_color="#333333", width=2)

    plt.axis("off")
    plt.tight_layout()
    plt.savefig(out_file, dpi=300, facecolor='white')  # Ensure white background
    plt.close()
    print(f"Saved graph → {out_file}")


# ===================================================================
# QUESTION 1 — ATTENTION IN TRANSFORMERS
# ===================================================================

nodes_q1 = [
    "Self-Attention Mechanism",
    "Dot-Product Attention",
    "Attention Score Scaling",
    "Multi-Head Attention",
    "Sequential Encoder-Decoder Model for Sequence-to-Sequence Tasks"
]

# Reasonable conceptual dependencies
edges_q1 = [
    ("Self-Attention Mechanism", "Dot-Product Attention"),
    ("Dot-Product Attention","Multi-Head Attention"),
    ("Dot-Product Attention", "Attention Score Scaling"),
    ("Self-Attention Mechanism", "Multi-Head Attention"),
    ("Multi-Head Attention", "Attention Score Scaling"),
    ("Multi-Head Attention", "Sequential Encoder-Decoder Model for Sequence-to-Sequence Tasks")
]

draw_graph(nodes_q1, edges_q1, "static/kg_q1.png")

# ===================================================================
# QUESTION 2 — CLIP EXPLANATION
# ===================================================================

nodes_q2 = [
    "CLIP Image Encoder",
    "CLIP Text Encoder",
    "Contrastive Loss (CLIP)",
    "Dot-Product Similarity",
    "One-Shot Classification"
]

edges_q2 = [
    ("CLIP Image Encoder", "Dot-Product Similarity"),
    ("CLIP Text Encoder", "Dot-Product Similarity"),
    ("Dot-Product Similarity", "Contrastive Loss (CLIP)"),
    ("Contrastive Loss (CLIP)", "One-Shot Classification"),
    ("CLIP Image Encoder", "Contrastive Loss (CLIP)"),
    ("CLIP Text Encoder", "Contrastive Loss (CLIP)")
]

draw_graph(nodes_q2, edges_q2, "static/kg_q3.png")

# ===================================================================
# QUESTION 3 — VARIATIONAL LOWER BOUND & JENSEN'S INEQUALITY
# ===================================================================

nodes_q3 = [
    "Jensen's Inequality",
    "Evidence Lower Bound (ELBO)",
    "Variational Inference",
    "Variational Autoencoder Loss",
    "Bias-Variance Tradeoff"
]

edges_q3 = [
    ("Jensen's Inequality", "Evidence Lower Bound (ELBO)"),
    ("Evidence Lower Bound (ELBO)", "Variational Inference"),
    ("Evidence Lower Bound (ELBO)", "Variational Autoencoder Loss"),
    ("Variational Inference", "Variational Autoencoder Loss"),
    ("Variational Autoencoder Loss", "Bias-Variance Tradeoff")
]

draw_graph(nodes_q3, edges_q3, "static/kg_q2.png")
```
